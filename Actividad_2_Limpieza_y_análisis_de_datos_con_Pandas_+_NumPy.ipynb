# **2. LIMPIEZA DE DATOS**

```python
# Eliminar la columna 'Animal code' y guardar el cambio en df
df= df.drop(columns=["Animal code"])
df.head()
```

## TRATAMIENTO DUPLICADOS EXACTOS

```python
#Cuántos registros duplicados existen?
print("SUMA DE DUPLICADOS: ",df.duplicated().sum())

print("Duplicados x columna: ")
df[df.duplicated()].head(50)
```

```python
##Borrar duplicados exactos

df = df.drop_duplicates()

df.shape
```

## TRATAMIENTO DUPLICADOS TECNICOS

```python
## Duplicados Tecnicos x columna

df["Animal type"].value_counts()
```

```python
## Edicion general de duplicados ANIMAL TYPE

df["Animal type"] = (
    df["Animal type"]
    .str.strip()
    .str.lower()
    .str.title()
)

df["Animal type"].value_counts()
```

```python
## Remplazar datos repetidos

df["Animal type"] = (
    df["Animal type"]
    .str.replace("Red Squirrell", "Red Squirrel", regex=False)
    .str.replace("Red Squirel", "Red Squirrel", regex=False)
    .str.replace("Ledgehod","Hedgehog",regex=False)
    .str.replace("Wedgehod","Hedgehog",regex=False)
    .str.replace("Lynx?","Lynx",regex=False)
    .str.replace("European Bison™","European Bison",regex=False)
    .str.replace("European Bisson","European Bison",regex=False)
)
df["Animal type"] = df["Animal type"].replace({"European Buster": "European Bison"})

df['Animal type'].value_counts()
```

```python
## columna Country

df["Country"].value_counts()
```

```python
# 1. Pasar todo a mayúsculas y quitar espacios
df["Country"] = df["Country"].str.strip().str.upper()

# 2. Reemplazar SOLO valores exactos usando un diccionario (evita que se dupliquen palabras)
df["Country"] = df["Country"].replace({
    "PL": "POLAND",
    "HU": "HUNGARY",
    "HUNGRY": "HUNGARY",
    "DE": "GERMANY",
    "CZ": "CZECH REPUBLIC",
    "CC": "CZECH REPUBLIC",
    "CZECH": "CZECH REPUBLIC",
    "AUSTRALIA": "AUSTRIA"  # En la base esos 4 registros tienen coordenadas de Austria
})

# 3. Dar formato de título (Primera letra en mayúscula)
df["Country"] = df["Country"].str.title()

# 4. Verificar que quedó limpio
df["Country"].value_counts()
```

```python
df['Gender'].value_counts()
```

```python
df["Gender"] = (
    df["Gender"]
    .str.strip()
    .str.lower()
    .str.title()
)

df["Gender"].value_counts()
```

## Numericos invalidos

```python
# Corregir tipos de datos numéricos
cols_numericas = ["Weight kg", "Body Length cm", "Latitude", "Longitude"]
for col in cols_numericas:
    df[col] = pd.to_numeric(df[col], errors="coerce")

df.tail()
```

```python
df['Body Length cm'].dtypes
```

#**VALORES FALTANTES**

```python
df.isna().sum()
```

```python
# 1. Rescatar los 8 registros sin especie deduciéndolos por su peso (< 0.38 kg son ardillas rojas)
condicion_ardilla = (df["Animal type"].isnull()) & (df["Weight kg"] < 0.38)
df.loc[condicion_ardilla, "Animal type"] = "Red Squirrel"

# Eliminar cualquier otra fila si quedara sin especie
df = df.dropna(subset=["Animal type"]).copy()

# 2. Rellenar los 4 vacíos de 'Gender' y los 795 vacíos de 'Animal name' con categorías lógicas
df["Gender"] = df["Gender"].fillna("Not Determined")
df["Animal name"] = df["Animal name"].fillna("Sin nombre")

# 3. Verificar que ya no quedan nulos en texto
df.isna().sum()
```

```python
# Resumen estadístico por especie para detectar valores imposibles
df.groupby("Animal type")[["Weight kg", "Body Length cm"]].describe()
```

```python
# PASO 1: Convertir valores imposibles a vacío (pd.NA) para que no dañen el cálculo

df.loc[df["Weight kg"] <= 0, "Weight kg"] = pd.NA
df.loc[(df["Animal type"] == "Lynx") & (df["Weight kg"] > 50), "Weight kg"] = pd.NA

df.loc[df["Body Length cm"] <= 0, "Body Length cm"] = pd.NA
df.loc[(df["Animal type"] == "Hedgehog") & (df["Body Length cm"] > 50), "Body Length cm"] = pd.NA

df.loc[(df["Latitude"] < 35) | (df["Latitude"] > 72), ["Latitude", "Longitude"]] = pd.NA
df.head()
```

```python
# PASO 2: Rellenar TODOS los vacíos (los que venían vacíos + los que acabamos de limpiar) con la MEDIANA de cada especie
df["Weight kg"] = df["Weight kg"].fillna(
    df.groupby("Animal type")["Weight kg"].transform("median")
)

df["Body Length cm"] = df["Body Length cm"].fillna(
    df.groupby("Animal type")["Body Length cm"].transform("median")
)
df['Body Length cm'].value_counts()
```

```python
# PASO 3: Verificar que ya no hay nulos ni negativos en peso y tamaño
print("Nulos restantes en peso y tamaño:")
print(df[["Weight kg", "Body Length cm"]].isnull().sum())

df.groupby("Animal type")[["Weight kg", "Body Length cm"]].describe()
```

```python
# Eliminar duplicados lógicos que hayan quedado tras limpiar textos, nulos y fechas
print("Duplicados finales encontrados:", df.duplicated().sum())
df = df.drop_duplicates().copy()

# Reiniciar el índice (para que las filas queden numeradas 0, 1, 2... sin saltos)
df = df.reset_index(drop=True)
df.head(12)
```

# **3. NUMPY**

```python
# 1. Crear una variable nueva con np.where() (ej. clasificar animales según su peso)
df["Tamaño_categoria"] = np.where(df["Weight kg"] > 10, "Grande/Mediano", "Pequeño")

# 2. Medida estadística con NumPy (usando nanmean por si quedara algún nulo)
promedio_largo = np.nanmean(df["Body Length cm"])
print("Promedio general de longitud corporal (cm):", round(promedio_largo, 2))
```

### Explicación de los resultados con NumPy
* **Variable `Tamaño_categoria` (`np.where()`):** Permite clasificar cada animal observado en `"Grande/Mediano"` (peso superior a 10 kg, agrupando bisontes y linces) o `"Pequeño"` (peso igual o menor a 10 kg, agrupando erizos y ardillas), facilitando el análisis por escala corporal.
* **Promedio general de longitud (`np.nanmean()`):** El promedio general de longitud corporal de toda la muestra es de **42.5 cm**. Este valor refleja el alto peso estadístico de las especies pequeñas (ardillas y erizos, que miden alrededor de 20 cm y representan más del 83 % de los registros), elevado por la presencia de especies mucho más largas como el bisonte europeo (hasta 350 cm).


#**4. ANÁLISIS CON GROUPBY()**

```python
# Análisis 1: Número de observaciones por tipo de animal
print("\n1. Observaciones por especie:")
print(df.groupby("Animal type")["Country"].count())

# Análisis 2: Peso promedio según el tipo de animal
print("\n2. Peso promedio (kg) por especie:")
print(df.groupby("Animal type")["Weight kg"].mean())

# Análisis 3: Longitud corporal promedio según el país
print("\n3. Longitud promedio (cm) por país:")
print(df.groupby("Country")["Body Length cm"].mean())

# Análisis 4: Peso promedio por especie y género
print("\n4. Peso promedio (kg) por especie y género:")
print(df.groupby(["Animal type", "Gender"])["Weight kg"].mean())
```

#**5. Interpretación de resultados**

1. **Predominio de especies pequeñas en las observaciones:**
La ardilla roja (Red Squirrel) y el erizo (Hedgehog) concentran la gran mayoría de los registros del estudio, con 421 y 278 observaciones respectivamente (sumando el 83.9 % del total), frente a solo 71 linces y 63 bisontes europeos. Esto indica que los roedores y pequeños mamíferos son mucho más abundantes o fáciles de avistar en las zonas muestreadas que los grandes mamíferos.
2. **Diferencia extrema en el peso corporal según la especie:** El bisonte europeo (European Bison) presenta el peso promedio más alto con **588.40 kg**, superando por más de 27 veces al lince (21.17 kg) y contrastando fuertemente con el erizo **(0.80 kg)** y la ardilla roja (0.30 kg). Esto evidencia la amplia diversidad de escala biológica dentro del ecosistema analizado, y confirma por qué era indispensable imputar los datos faltantes agrupando por especie y no con un promedio general.
3. **Variación geográfica en el tamaño promedio de los animales observados:** Polonia registra la mayor longitud corporal promedio con 74.01 cm, seguida de República Checa (52.01 cm) y Eslovaquia (43.39 cm), mientras que Alemania presenta el promedio más bajo con 20.11 cm. Esto significa que en Polonia se concentra una mayor proporción de avistamientos de fauna de gran tamaño (como bisontes europeos), mientras que en Alemania y Austria (24.35 cm) los registros corresponden casi exclusivamente a especies pequeñas como ardillas y erizos.
4. **Dimorfismo sexual marcado en mamíferos grandes frente a especies pequeñas:** En el bisonte europeo (European Bison), los machos alcanzan un peso promedio de 728.00 kg frente a los 481.11 kg de las hembras (una diferencia de más de 246 kg), mientras que en la ardilla roja (Red Squirrel) ambos sexos mantienen un peso prácticamente idéntico (0.297 kg). Esto demuestra que el dimorfismo sexual en masa corporal es muy pronunciado en los grandes herbívoros de la muestra, pero inexistente en los pequeños roedores.
